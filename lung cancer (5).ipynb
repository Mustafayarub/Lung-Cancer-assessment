```python
X_train,X_test,y_train,y_test=train_test_split(X,y,random_state=3,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
model.fit(X_train,y_train)
```

```python
prediction=model.predict(X_test)
prediction
```

```python
from sklearn import metrics
```

```python
accuracy = metrics.accuracy_score(y_test,prediction)
accuracy
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='binary')
recall = recall_score(y_test, y_pred, average='binary')
f1 = f1_score(y_test, y_pred, average='binary')
print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-score: {f1:.2f}")
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))
```

```python
import customtkinter as ctk
from tkinter import filedialog, messagebox
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
import numpy as np

# --- 1. إعدادات التصميم ---
ctk.set_appearance_mode("Dark")
ctk.set_default_color_theme("dark-blue")

# لوحة الألوان
COLOR_BG = "#15191E"
COLOR_PANEL = "#21262D"
COLOR_ACCENT = "#3A86FF"
COLOR_GLOW = "#00F0FF"
COLOR_TEXT = "#E6EDF3"
COLOR_SELECTED = "#238636"

class ProAnalyticsApp(ctk.CTk):
    def __init__(self):
        super().__init__()
        self.title("🧬 LUNG CANCER ANALYTICS | PRO VERSION")
        self.geometry("1500x950")
        self.configure(fg_color=COLOR_BG)

        self.df = None
        self.model = None
        self.encoders = {} 
        self.feature_cols = []
        self.input_widgets = {} # لتخزين أدوات الإدخال

        self.setup_ui()

    def setup_ui(self):
        self.grid_columnconfigure(1, weight=1)
        self.grid_rowconfigure(0, weight=1)

        # القائمة الجانبية
        self.sidebar = ctk.CTkFrame(self, width=280, corner_radius=0, fg_color=COLOR_PANEL)
        self.sidebar.grid(row=0, column=0, sticky="nsew")
        ctk.CTkLabel(self.sidebar, text="🧬Dr.Suhad-AI", font=("Impact", 30), text_color=COLOR_GLOW).pack(pady=40)
        
        self.create_glow_btn("📂 IMPORT DATA", self.load_data)
        self.create_glow_btn("🧠 TRAIN SYSTEM", self.train_model)
        self.create_glow_btn("📊 VISUALIZE", self.show_analytics)
        self.create_glow_btn("⚡ PREDICT RISK", self.show_prediction)
        
        self.status_lbl = ctk.CTkLabel(self.sidebar, text="⚫ System Idle", font=("Consolas", 12), text_color="grey")
        self.status_lbl.pack(side="bottom", pady=20)

        # المحتوى الرئيسي
        self.main_area = ctk.CTkFrame(self, fg_color=COLOR_BG, corner_radius=0)
        self.main_area.grid(row=0, column=1, sticky="nsew", padx=20, pady=20)
        
        self.frames = {}
        self.frames["analytics"] = ctk.CTkFrame(self.main_area, fg_color="transparent")
        self.setup_analytics_ui(self.frames["analytics"])
        
        self.frames["prediction"] = ctk.CTkScrollableFrame(self.main_area, fg_color="transparent")
        
        # عرض الشاشة الافتراضية
        self.show_analytics()

    def create_glow_btn(self, text, cmd):
        btn = ctk.CTkButton(self.sidebar, text=text, command=cmd, 
                            font=("Segoe UI", 14, "bold"), height=55, corner_radius=8,
                            fg_color="transparent", border_width=2, border_color="#30363D",
                            hover_color=COLOR_ACCENT, text_color="white", anchor="w")
        btn.pack(pady=8, padx=20, fill="x")

    def switch_frame(self, name):
        for f in self.frames.values(): f.pack_forget()
        self.frames[name].pack(fill="both", expand=True)

    def show_analytics(self): self.switch_frame("analytics")
    def show_prediction(self): self.switch_frame("prediction")

    # ------------------------------------------------------------------
    # MODEL LOGIC
    # ------------------------------------------------------------------
    def load_data(self):
        path = filedialog.askopenfilename(filetypes=[("CSV", "*.csv")])
        if path:
            try:
                self.df = pd.read_csv(path)
                # تنظيف أسماء الأعمدة لإزالة المسافات الزائدة
                self.df.columns = self.df.columns.str.strip().str.upper()
                
                # تعديل الجندر ليكون مقروءاً إذا كان أرقاماً
                if "GENDER" in self.df.columns:
                    # افتراض أن 1 ذكر و 2 أنثى حسب البيانات الشائعة، يمكن تعديلها
                    self.df["GENDER"] = self.df["GENDER"].replace({1:"Male", 2:"Female"})
                
                self.status_lbl.configure(text="✅ Data Loaded", text_color=COLOR_GLOW)
                messagebox.showinfo("Success", "Dataset loaded successfully!")
                
                # تحديث واجهة التوقع فور تحميل البيانات
                self.setup_prediction_ui()
                
            except Exception as e:
                messagebox.showerror("Error", f"Bad File: {e}")

    def train_model(self):
        if self.df is None: 
            messagebox.showwarning("Warning", "Please import data first!")
            return
            
        self.status_lbl.configure(text="⚙️ Training...", text_color="orange")
        self.update()
        
        try:
            df_train = self.df.copy()
            target = "LUNG_CANCER"
            
            if target not in df_train.columns:
                messagebox.showerror("Error", f"Column '{target}' not found in dataset!")
                return

            X = df_train.drop(target, axis=1)
            y = df_train[target]
            
            self.feature_cols = X.columns.tolist()
            self.encoders = {}
            
            # معالجة البيانات النصية
            for col in X.columns:
                if X[col].dtype == 'object' or X[col].nunique() < 10:
                    le = LabelEncoder()
                    X[col] = le.fit_transform(X[col].astype(str))
                    self.encoders[col] = le
            
            self.target_enc = LabelEncoder()
            y = self.target_enc.fit_transform(y)
            
            self.model = RandomForestClassifier(n_estimators=200, max_depth=10, random_state=42)
            self.model.fit(X, y)
            
            self.status_lbl.configure(text="🟢 Model Ready", text_color="#00FF00")
            messagebox.showinfo("AI Core", "Training Complete & Optimized.")
            
        except Exception as e:
            self.status_lbl.configure(text="🔴 Error", text_color="red")
            messagebox.showerror("Train Error", str(e))

    # ------------------------------------------------------------------
    # ANALYTICS UI
    # ------------------------------------------------------------------
    def setup_analytics_ui(self, parent):
        ctrl = ctk.CTkFrame(parent, fg_color=COLOR_PANEL, height=60, corner_radius=10)
        ctrl.pack(fill="x", pady=(0, 15))
        
        self.chart_type = ctk.CTkComboBox(ctrl, width=300, height=40, font=("Arial", 14),
                                          values=[
                                              "1. Gender Distribution (Donut Chart)",
                                              "2. Age vs Cancer (Violin Plot)",
                                              "3. Smoking & Cancer Risk (Bar)",
                                              "4. Symptoms Heatmap (Correlation)",
                                              "5. Alcohol vs Age (Scatter)"
                                          ], border_color=COLOR_ACCENT, button_color=COLOR_ACCENT)
        self.chart_type.pack(side="left", padx=20, pady=10)
        
        ctk.CTkButton(ctrl, text="RENDER CHART 🎨", command=self.render_chart, 
                      font=("Arial", 12, "bold"), fg_color=COLOR_ACCENT, hover_color=COLOR_GLOW, height=40).pack(side="left")
        
        self.plot_frame = ctk.CTkFrame(parent, fg_color=COLOR_PANEL, corner_radius=10)
        self.plot_frame.pack(fill="both", expand=True)
        
        plt.style.use('dark_background')
        self.fig = plt.Figure(figsize=(6, 5), dpi=100)
        self.fig.patch.set_facecolor(COLOR_PANEL)
        
        self.canvas = FigureCanvasTkAgg(self.fig, master=self.plot_frame)
        self.canvas.get_tk_widget().pack(fill="both", expand=True, padx=10, pady=10)

    def render_chart(self):
        if self.df is None: return
        sel = self.chart_type.get()
        
        # تنظيف الرسم البياني السابق بشكل صحيح
        self.fig.clear()
        ax = self.fig.add_subplot(111)
        ax.set_facecolor(COLOR_PANEL)
        
        viz_df = self.df.copy()

        # تحويل القيم الرقمية 1 و 2 إلى نصوص للعرض الجميل
        for col in viz_df.columns:
            if col not in ["AGE", "GENDER"] and viz_df[col].dtype != 'object':
                if viz_df[col].nunique() <= 2:
                    viz_df[col] = viz_df[col].replace({1: 'NO', 2: 'YES'})

        viz_df['AGE'] = pd.to_numeric(viz_df['AGE'], errors='coerce')
        palette_bi = ["#00C853", "#D50000"]

        try:
            if "1." in sel:  # Gender
                counts = viz_df['GENDER'].value_counts()
                ax.pie(counts, labels=counts.index, autopct='%1.1f%%', colors=["#3A86FF", "#FF006E"], wedgeprops={'width': 0.4})
                ax.set_title("Gender Distribution")

            elif "2." in sel:  # Age Violin
                sns.violinplot(x="LUNG_CANCER", y="AGE", data=viz_df, ax=ax, palette="muted")
                ax.set_title("Age Density Analysis")

            elif "3." in sel:  # Smoking
                sns.countplot(x="SMOKING", hue="LUNG_CANCER", data=viz_df, ax=ax, palette=palette_bi)
                ax.set_title("Smoking Impact on Cancer")

            elif "4." in sel:  # Heatmap
                num_df = self.df.copy()
                le = LabelEncoder()
                for c in num_df.columns:
                    if num_df[c].dtype == 'object': num_df[c] = le.fit_transform(num_df[c])
                sns.heatmap(num_df.corr(), ax=ax, cmap="icefire", cbar=True)
                ax.set_title("Feature Correlation")

            elif "5." in sel:  # Alcohol Scatter
                sns.stripplot(x="ALCOHOL CONSUMING", y="AGE", hue="LUNG_CANCER", data=viz_df, ax=ax, palette=palette_bi, alpha=0.7, jitter=True)
                ax.set_title("Alcohol vs Age Clusters")

            # تنسيق المظهر العام
            ax.spines['top'].set_visible(False)
            ax.spines['right'].set_visible(False)
            ax.tick_params(colors='white', labelsize=9)
            ax.xaxis.label.set_color('white')
            ax.yaxis.label.set_color('white')
            ax.title.set_color(COLOR_GLOW)
            
            self.canvas.draw()
            
        except Exception as e:
            ax.text(0.5, 0.5, f"Error: {str(e)}", ha='center', color='red')
            self.canvas.draw()

    # ------------------------------------------------------------------
    # PREDICTION UI (THE FIXED PART)
    # ------------------------------------------------------------------
    def setup_prediction_ui(self):
        # تنظيف أي عناصر قديمة
        for widget in self.frames["prediction"].winfo_children(): 
            widget.destroy()
            
        if self.df is None: return
        
        # --- Header ---
        header = ctk.CTkFrame(self.frames["prediction"], fg_color="transparent")
        header.pack(fill="x", pady=20)
        ctk.CTkLabel(header, text="PATIENT DIAGNOSIS FORM", font=("Segoe UI", 24, "bold"), text_color="white").pack()
        ctk.CTkLabel(header, text="Select YES or NO for each symptom", font=("Arial", 12), text_color="grey").pack()

        self.input_widgets = {}
        features = [col for col in self.df.columns if col != "LUNG_CANCER"]
        
        # --- Grid Layout Container ---
        grid_container = ctk.CTkFrame(self.frames["prediction"], fg_color="transparent")
        grid_container.pack(fill="both", expand=True, padx=40)

        for i, feat in enumerate(features):
            row = i // 2
            col = i % 2
            
            # البطاقة الخاصة بكل خاصية
            card = ctk.CTkFrame(grid_container, fg_color=COLOR_PANEL, corner_radius=12, border_width=1, border_color="#333")
            card.grid(row=row, column=col, sticky="ew", padx=10, pady=10, ipady=5)
            
            # ضبط الأعمدة لتتمدد
            grid_container.grid_columnconfigure(0, weight=1)
            grid_container.grid_columnconfigure(1, weight=1)
            
            lbl = ctk.CTkLabel(card, text=f" {feat}", font=("Arial", 13, "bold"), text_color="#DDD")
            lbl.pack(anchor="w", padx=15, pady=(10, 5))

            # استخراج القيم الفريدة للعمود
            unique_vals = sorted(self.df[feat].unique().astype(str).tolist())
            
            # --- منطق الأزرار (Toggle Buttons) ---
            if len(unique_vals) <= 2:
                if feat == "GENDER":
                    display_vals = ["Male", "Female"]
                else:
                    display_vals = ["NO", "YES"]
                
                # [FIX IMPORTANT] إنشاء خريطة تربط العرض بالقيمة الحقيقية
                # مثال: {"NO": "1", "YES": "2"}
                # نأخذ القيمة الأولى والأخيرة لضمان التوافق
                safe_mapping = {display_vals[0]: unique_vals[0], display_vals[1]: unique_vals[-1]}

                inp = ctk.CTkSegmentedButton(card, values=display_vals, font=("Arial", 12, "bold"), height=35,
                                            fg_color="#0D1117", selected_color=COLOR_ACCENT, selected_hover_color=COLOR_GLOW,
                                            unselected_color="#0D1117", unselected_hover_color="#333", text_color="white")
                inp.set(display_vals[0])
                inp.pack(fill="x", padx=15, pady=(0, 15))
                
                # حفظ الأداة مع الخريطة الخاصة بها
                self.input_widgets[feat] = {"type": "toggle", "widget": inp, "mapping": safe_mapping}
            
            # --- منطق الإدخال الرقمي (Entry) ---
            else:
                frame_inp = ctk.CTkFrame(card, fg_color="transparent")
                frame_inp.pack(fill="x", padx=15, pady=5)
                inp = ctk.CTkEntry(frame_inp, height=35, border_color="#444", justify="center")
                try: inp.insert(0, str(int(self.df[feat].mean())))
                except: pass
                inp.pack(fill="x")
                self.input_widgets[feat] = {"type": "entry", "widget": inp}

        # زر التشخيص
        btn_predict = ctk.CTkButton(self.frames["prediction"], text="RUN DIAGNOSIS PROTOCOL ⚡", 
                                    command=self.run_prediction, height=60, font=("Impact", 18), 
                                    fg_color=COLOR_ACCENT, hover_color=COLOR_GLOW, corner_radius=15)
        btn_predict.pack(pady=40, padx=100, fill="x")

    def run_prediction(self):
        if self.model is None:
            messagebox.showerror("Error", "Please Train the Model First!")
            return
        try:
            input_row = []
            for col in self.feature_cols:
                item = self.input_widgets[col]
                widget = item["widget"]
                w_type = item["type"]
                val = widget.get()

                if w_type == "toggle":
                    # [FIX IMPORTANT] استخدام الخريطة المحفوظة لجلب القيمة الحقيقية
                    raw_val = item["mapping"][val]
                    
                    if col in self.encoders:
                        final_val = self.encoders[col].transform([str(raw_val)])[0]
                    else:
                        try: final_val = int(raw_val)
                        except: final_val = float(raw_val)
                else:
                    final_val = float(val)
                
                input_row.append(final_val)

            # تنفيذ التوقع
            probs = self.model.predict_proba([input_row])[0]
            # الاحتمال الثاني عادة هو Positive (Cancer)
            risk_score = probs[1] if len(probs) > 1 else probs[0]
            self.show_result_popup(risk_score)

        except Exception as e:
            messagebox.showerror("Prediction Error", f"Check inputs.\nDetails: {e}")

    def show_result_popup(self, risk):
        top = ctk.CTkToplevel(self)
        top.geometry("450x450")
        top.title("DIAGNOSTIC REPORT")
        top.configure(fg_color="#111")
        top.attributes("-topmost", True)
        
        if risk > 0.5:
            color = "#FF003C"
            status = "HIGH RISK DETECTED"
            icon = "⚠️"
        else:
            color = "#00FF66"
            status = "LOW RISK / NEGATIVE"
            icon = "✅"
            
        ctk.CTkLabel(top, text=icon, font=("Arial", 60)).pack(pady=(40, 10))
        ctk.CTkLabel(top, text=status, font=("Impact", 28), text_color=color).pack(pady=10)
        
        fr = ctk.CTkFrame(top, fg_color="#222", corner_radius=20)
        fr.pack(pady=20, padx=40, fill="x")
        
        ctk.CTkLabel(fr, text=f"CONFIDENCE: {risk*100:.1f}%", font=("Consolas", 16, "bold"), text_color="white").pack(pady=10)
        
        prog = ctk.CTkProgressBar(fr, height=15, progress_color=color)
        prog.pack(fill="x", padx=20, pady=(0, 20))
        prog.set(risk)
        
        ctk.CTkButton(top, text="CLOSE REPORT", command=top.destroy, fg_color="#333", hover_color="#555", width=200).pack(pady=20)

if __name__ == "__main__":
    app = ProAnalyticsApp()
    app.mainloop()
```
